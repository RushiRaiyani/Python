### 01) WAP to check whether the given string is palindrome or not.

```python
a=input('Enter a string:')
reverse=a[::-1]
if a==reverse:
    print('Given string is palindrome')
else:
    print('Given string is not palindrome')
```

### 02) WAP to reverse the words in the given string.

```python
a=input('Enter a string:')
temp=a.split(' ')[::-1]
ans1=' '.join(temp)
print('reversed word string is:',ans1)
temp2=[];
temp=a.split(' ')
for i in temp:
    temp2.append(i[::-1])
ans2=' '.join(temp2)
print('reversed word letter string is:',ans2)
```

### 05) WAP to print even length word in string.

```python
a=input('Enter a string:')
temp=a.split(' ')
ansList=[]
for i in temp:
    if(len(i)%2==0):
        ansList.append(i)
ans=''.join(ansList)
print('even length word in string is:',ans)
```

### 06) WAP to count numbers of vowels in given string.

```python
a=input('Enter a string:')
count=0
for i in a:
    if(i=='a' or i=='e' or i=='i' or i=='o' or i=='u'):
        count=count+1
print('Number of vowel are:',count)
```

### 07) WAP to capitalize the first and last character of each word in a string.

```python
a=input('Enter a string:')
tempList=a.split(' ')
ansList=[];
for i in tempList:
    ansList.append(i[0].upper()+i[1:len(i)-1]+i[-1].upper())
ans=' '.join(ansList)
print('String after making capitalize of first and last character of each word is:',ans)
```

### 10) : Display credit card number. 
### card no. : 1234 5678 9012 3456
### display as  : **** **** **** 3456

```python
card_number = "1234 5678 9012 3456"
masked_card = ""
for i in range(len(card_number)):
    if i < len(card_number) - 4 and card_number[i] != " ":
        masked_card += "*"
    else:
        masked_card += card_number[i]

print("Card No.:", masked_card)
```

### 12) : Rearrange the given string. First lowercase then uppercase alphabets.
### input : EHlsarwiwhtwMV
### output : lsarwiwhtwEHMV

```python
s=input('Enter a string A:')
ans=''
ans2=''
for i in s:
    if(i.islower()):
        ans=ans+i
    else:
        ans2=ans2+i
final=ans+ans2
print('New string is:',final)
```
