# if..else..

### 01) WAP to check whether the given number is positive or negative.

```python
number=float(input('Enter a number:'))
if number>0:
    print(f'{number} is positive')
elif number==0:
    print('Number is 0')
else:
    print(f'{number} is negative')
```

### 02) WAP to check whether the given number is odd or even.

```python
number=int(input('Enter a number'))
if number%2==0:
    print(f'{number} is even')
else:
    print(f'{number} is odd')
```

### 03) WAP to find out largest number from given two numbers using simple if and ternary operator.

```python
a=float(input('Enter a number A:'))
b=float(input('Enter a number B:'))
print('using if else')
if a>b:
    print(f'{a} is greater than {b}')
else:
    print(f'{b} is greater than {a}')
print('using ternary operator')
max = f"{a} is maximum" if a > b else f"{b} is maximum"
print(max)
```

### 05) WAP to check whether the given year is leap year or not.
[If a year can be divisible by 4 but not divisible by 100 then it is leap year but if it is divisible by 400 then it is leap year]

```python
y=int(input('Enter a year:'))
if(y%4==0 and y%100!=0):
    print(f'{y} is a leap year')
elif(y%400==0):
    print(f'{y} is a leap year')
else:
    print(f'{y} is not leap year')
```

### 07) WAP to implement simple calculator which performs (add,sub,mul,div) of two no. based on user input.

```python
a=int(input('Enter a number A:'))
b=int(input('Enter a number B:'))
sign=input('Enter a operation:')
match sign:
    case '+':
        print(f'{a} {sign} {b} is', (a+b))
    case '-':
        print(f'{a} {sign} {b} is', (a-b))
    case '*':
        print(f'{a} {sign} {b} is', (a*b))
    case '/':
        print(f'{a} {sign} {b} is', (a/b))
    case _:
        print('Enter a correct sign')
```

### 08) WAP to read marks of five subjects. Calculate percentage and print class accordingly. 
Fail below 35 </br>
Pass Class between 35 to 45 </br>
Second Class</br>
between 45 to 60</br>
First Class between 60 to 70</br> 
Distinction if more than 70

```python
ans=0
for i in range(1, 6):
    a=int(input(f'Enter marks of subject {i}:'))
    ans=ans+a
percent=(ans*100)/500
if (percent>70):
    print('Distinction')
elif(percent>60):
    print('First class')
elif(percent>45):
    print('Second class')
elif(percent>=35):
    print('Pass class')
elif(percent<35):
    print('Sorry you are Fail')
else:
    print('Enter correct input')
```

### 09) Three sides of a triangle are entered through the keyboard, WAP to check whether the triangle is isosceles, equilateral, scalene or right-angled triangle.

```python
a=float(input('Enter a side A:'))
b=float(input('Enter a side B:'))
c=float(input('Enter a side C:'))
if(a==b and a==c):
    print('Triangle is Equilateral')
elif(a==b or b==c or a==c):
    print('Triangle is isosceles')
elif(((a*a)+(b*b)==(c*c)) or ((a*a)+(c*c)==(b*b)) or ((c*c)+(b*b)==(a*a))):
    print('Triangle is right-angled')
else:
    print('Triangle is scalene')
```
