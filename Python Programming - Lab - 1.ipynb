### 09) WAP to find the distance between two points in 2-D space.

```python
import math
a=float(input('Enter x coordinate of point A'))
b=float(input('Enter y coordinate of point A'))
c=float(input('Enter x coordinate of point B'))
d=float(input('Enter y coordinate of point B'))
e=((a-b)**2)+((c-d)**2)
print('Distance between point A and B is', (math.sqrt(e)))
```

### 11) WAP to print sum of square of n natural numbers.

```python
a=int(input('Enter number of terms'))
ans,j=0,1
for i in range(0,a):
    ans=ans+(j*j)
    j=j+1
print(ans)
```
