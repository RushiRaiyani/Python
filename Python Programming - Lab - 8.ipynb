### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def Prime(n):
    isPrime=True
    for i in range(2,int(n/2)+1):
        if n%i==0:
            return 0
    return 1

n1=int(input("Enter the value of n"))
print(Prime(n1))
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def PrimeNum(s,e):
    li1=[]
    for j in range (s+1,e):
        isPrime=True
        for i in range(2,int(j/2)+1):
            if j%i==0:
                isPrime=False
                break
        if isPrime:
            li1.append(j)
    return li1
start=int(input("Enter the starting number"))
end=int(input("Enter the ending number"))
print(PrimeNum(start,end))
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def Palindrome(s):
    rev=s[::-1]
    if(s==rev):
        return True
    return False

s= input("Enter the String:")
print(Palindrome(s))
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
# starts with 1 like 1,1,2,3,5,8,13...
def fibonacci(n):
    if n==0:
        return 0
    if n==1:
        return 1

    return fibonacci(n-1)+fibonacci(n-2)

print(fibonacci(6))
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
def SumOfScores(l10):
    sum=0
    for i in l10:
        if(i%10==0):
            sum=sum+i
    return sum

scores= [200, 456, 300, 100, 234, 678]
print(SumOfScores(scores))
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
def InvertedDict(dict):
    newDict={}
    for i in dict:
        newDict[dict[i]]=i
    return newDict
    
d11={}
k=input("Enter the Key:")
while k!='-1':
    val=int(input("Enter the Value:"))
    d11[k]=val
    k=input("Enter the Key:")

print(d11)
print(InvertedDict(d11))
```

### 12) Write a function to check whether the given string is Pangram or not.
#### hint: Pangram is a string containing all the characters a-z atlest once.
#### "the quick brown fox jumps over the lazy dog" is a Pangram string.

```python
def isPangram(s):
    for i in range(97,123):
        if chr(i) not in s:
            return "Not Pangram String"
    return "Pangram String"
string = input("Enter the string:")
print(isPangram(string))
```

### 13) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def numOfUpperLower(s):
    li=[0,0]
    for i in s:
        if i.isupper():
            li[0]=li[0]+1
        else:
            li[1]=li[1]+1
    return li

str13=input("Enter the String")
count=numOfUpperLower(str13)
print(f"upper={count[0]} lower={count[1]}")
```
