### 05) WAP to count the no. of lines, words and characters in a given text file.

```python
fp=open("Rashi.txt","r")
lines=(fp.readlines())

char=0
words=0
for i in lines:
    temp=i.split()
    words=words+len(temp)
    char=char+len(i)

print(f"lines={len(lines)}  words={words}  chars={char}")
fp.close()
```

### 08) WAP to create an UDF named frequency to count occurances of the specific word in a given text file.

```python
def Frequency(word):
    fp=open("Rashi.txt","r")
    count=0
    lines=fp.readlines()
    for i in lines:
        words=i.split()
        if word in words:
            count=count+1
    return count

word=input("Enter the word:")
print(f"{Frequency(word)}")
fp.close()
```

### 09) WAP to get the score of five subjects from the user, store them in a file. Fetch those marks and find the highest score.

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
import math
prime=[]
pcount=0
i=2
while 1:
    isPrime=True
    for j in range(2,math.isqrt(i)+1):
        if i%j==0:
            isPrime=False
            break
    if isPrime:
        pcount=pcount+1
        num=str(i)+'\n'
        prime.append(num)
                   
    if(pcount==100):
        break
    i=i+1

fp=open("primenumbers.txt","w")
fp.writelines(prime)
fp.close()
```
