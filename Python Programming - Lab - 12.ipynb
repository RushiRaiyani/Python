### 04) WAP to demonstrate the use of Pie chart.

### 05) WAP to demonstrate the use of Bar chart.

```python
x=[1,2,3,4,5]
y=[10,30,50,20,60]
c=['b','g','y','c','r']
l=['1st','2nd','3rd','4th','5th']
w=[0.5,0.7,0.8,0.6,0.9]
plt.title('Report')
for i in range(len(x)):
    plt.bar(x[i], y[i], color=c[i], label=l[i], width=w[i])

plt.legend()
plt.show()
```

### 08) WAP to display the value of each bar in a bar chart using Matplotlib.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 30, 50, 20, 60]
c = ['b', 'g', 'y', 'c', 'r']
l = ['1st', '2nd', '3rd', '4th', '5th']
w = [0.5, 0.7, 0.8, 0.6, 0.9]

plt.title('Report')
bars = plt.bar(x, y, color=c, label=l, width=w)

# Display the value of each bar on top of the bars
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # X position
        bar.get_height(),                  # Y position
        str(bar.get_height()),              # Text to display
        ha='center',                       # Horizontal alignment
        va='bottom'                        # Vertical alignment
    )

plt.show()
```
