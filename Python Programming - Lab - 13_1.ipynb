### 03) WAP to create Circle class with area and perimeter function to find area and perimeter of circle.

```python
class Circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        print(f'Area of circle is: {3.14*self.radius*self.radius}')
    def perimeter(self):
        print(f'Perimeter of circle is: {2*3.14*self.radius}')
radius=float(input('Enter radius of circle'))
circle=Circle(radius)
circle.area()
circle.perimeter()
```

### 05) Create a bank account class with methods to deposit, withdraw, and check balance.

```python
class BankAccount:
    def __init__(self,amount):
        self.amount=amount
    def deposit(self,depositAmount):
        self.amount=self.amount+depositAmount
    def withdraw(self,withdrawAmount):
        if self.amount-withdrawAmount<0:
            print('value too high to withdraw')
        else:
            self.amount=self.amount-withdrawAmount
    def display(self):
        print(f'Balance is: {self.amount}')

amt=int(input('Enter amount to add in account:'))
account=BankAccount(0)
account.deposit(amt)
withdraw=int(input('Enter amount to withdraw in account:'))
account.withdraw(withdraw)
account.display()
```

### 06) Create a class for managing inventory that includes attributes such as item name, price, quantity, and methods to add, remove, and update items.

```python
class Inventory:
    def __init__(self):
        self.items = {}  # Dictionary to store items
    
    def add(self, name, price, quantity):
        if name in self.items:
            print("Item already exists.")
        else:
            self.items[name] = {'price': price, 'quantity': quantity}
    
    def remove(self, name):
        if name in self.items:
            del self.items[name]
            print(f"Item '{name}' removed.")
        else:
            print("No such item found.")
    
    def update(self, name, price, quantity):
        if name in self.items:
            if price is not None:
                self.items[name]['price'] = price
            if quantity is not None:
                self.items[name]['quantity'] = quantity
            print(f"Item '{name}' updated.")
        else:
            print("No such item found.")
    
    def display(self):
        if not self.items:
            print("Inventory is empty.")
        else:
            for name, details in self.items.items():
                print(f"{name}: Price - {details['price']}, Quantity - {details['quantity']}")
    def getInput(self):
        name = input("Enter item name: ")
        price = int(input("Enter price of item: "))
        quantity = int(input("Enter quantity number: "))
        return name,price,quantity


inventory = Inventory()

l=inventory.getInput()
inventory.add(l[0], l[1], l[2])

l=inventory.getInput()
inventory.add(l[0], l[1], l[2])

remove=(input('Enter name of item to remove'))
inventory.remove(remove)
l=inventory.getInput()
inventory.update(l[0], l[1], l[2])
inventory.display()
```

### 09) Define Time class with hour and minute as data member. Also define addition method to add two time objects.

```python
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute
    
    def add(self, other):
        total_minutes = self.minute + other.minute
        extra_hours = total_minutes // 60
        new_minutes = total_minutes % 60
        new_hours = self.hour + other.hour + extra_hours
        return Time(new_hours, new_minutes)
    
    def display(self):
        print(f"{self.hour} hours, {self.minute} minutes")


time1 = Time(2, 45)
time2 = Time(1, 30)
time3 = time1.add(time2)
time3.display()
```
