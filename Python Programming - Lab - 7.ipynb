### 09) WAP to check whether two given strings are anagram or not using set.

```python
a=input('Enter a string 1:')
b=input('Enter a string 2:')
if len(a)!=len(b):
    print('not anagram')
else:
    d1={}
    d2={}
    for i in a:
        if i in d1:
            d1[i]=d1[i]+1
        else:
            d1.setdefault(i,0)
    for i in b:
        if i in d2:
            d2[i]=d2[i]+1
        else:
            d2.setdefault(i,0)
    anagram=True
    for i in d1:
        if d1[i]!=d2.get(i,0):
            anagram=False
            print('Not an anagram')
            break
    if anagram is True:
        print('anagram')
```

### 10) WAP to find common elements in three lists using set.

```python
a=int(input('Enter number of terms:'))
l1=[]
for i in range(a):
    k=int(input('Enter term:'))
    l1.append(k)

b=int(input('Enter number of terms:'))
l2=[]
for i in range(b):
    k=int(input('Enter term:'))
    l2.append(k)

c=int(input('Enter number of terms:'))
l3=[]
for i in range(c):
    k=int(input('Enter term:'))
    l3.append(k)


common_elements = set(l1) & set(l2) & set(l3)
print(f"Common elements: {common_elements}")
```

### 13) WAP to sort dictionary by key or value.

```python
a=int(input('Enter number of terms:'))
dic1={}
for i in range(a):
    k=int(input('Enter key:'))
    v=int(input('Enter value:'))
    dic1[k]=v

sorted_by_keys = {key: dic1[key] for key in sorted(dic1)}

print("Dictionary sorted by keys:", sorted_by_keys)
```

### 14) WAP to find the sum of all items (values) in a dictionary given by user. (Assume: values are numeric)

```python
a=int(input('Enter number of terms:'))
dic1={}
for i in range(a):
    k=int(input('Enter key:'))
    v=int(input('Enter value:'))
    dic1[k]=v
sum=0
for i in dic1:
    sum=sum+dic1[i]
print(sum)
```
