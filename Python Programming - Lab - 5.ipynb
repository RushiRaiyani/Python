# List

### 01) WAP to find sum of all the elements in a List.

```python
list=[1,4,10,-123,328,11]
sum=0
for i in list:
    sum=sum+i
print(f'Sum of all elements of list is:${sum}')
```

### 02) WAP to find largest element in a List.

```python
list=[1,2,23,12,5]
max=list[0]
for i in list:
    if i>max:
        max=i
print('Maximum element in list is:',max)
```

### 03) WAP to find the length of a List.

### 04) WAP to interchange first and last elements in a list.

```python
list=[1,2,3,4,5,6]
last=list.pop()
first=list.pop(0)
list.insert(0,last)
list.append(first)
print('New List is:',list)
```

### 05) WAP to split the List into two parts and append the first part to the end.

```python
list=[1,2,3,4,5,6]
list2=[]
for i in range (0,int(len(list)/2)):
    list2.append(list[i])
    list.pop(i)
list.extend(list2)
print('New List is:',list)
```

### 12) WAP to count occurrences of an element in a list.

### 14) WAP to extract elements with frequency greater than K.

```python
dic={}
list=[1,2,3,4,5,3,5,1,1]
freq=int(input('Enter frequency:'))
for i in list:
    dic.setdefault(i,0)
    dic[i]=dic[i]+1
for i in dic:
    if dic[i]>freq:
        print(i)
```

### 15) WAP to create a list of squared numbers from 0 to 9 with and without using List Comprehension.

```python
list1=[]
for i in range(0,10):
    list1.append(i*i)
list2=[i*i for i in range(0,10)]
print('List1 without comprehension:',list1)
print('List2 with comprehension',list2)
```

### 16) WAP to create a new list (fruit whose name starts with 'b') from the list of fruits given by user.

```python
list=["apple", "banana","mango", "blue-berry"]
list2=[]
for i in list:
    temp = [ch for ch in i]
    if temp[0]=='b' or temp[0]=='B':
        list2.append(i)
print('List containing B as starting letter in fruit name:',list2)
```

### 17) WAP to create a list of common elements from given two lists.

```python
list1=[1,1,1,3,2,3,4,5]
list2=[1,1,2,3,3,3,3,45,6]
list3=[]
for i in list1:
    if i in list2:
        list3.append(i)
        list2.remove(i)
print('common element list is:',list3)
```
