### 12) Calculate the area of a rectangle. 
### Include a Constructor, a method to calculate area named area() and a method named output() that prints the output and is invoked by area().
### Also define a class method that compares the two sides of reactangle. An object is instantiated only if the two sides are different; otherwise a message should be displayed : THIS IS SQUARE.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        area = self.length * self.width
        self.output(area)

    def output(self, area):
        print(f"Area of Rectangle: {area}")

    @classmethod
    def create_rectangle(cls, length, width):
        if length == width:
            print("THIS IS SQUARE.")
        else:
            return cls(length, width)


length = int(input("Enter length: "))
width = int(input("Enter width: "))

rect = Rectangle.create_rectangle(length, width)

if rect:
    rect.area()
```

### 14) Create a class Profit that has a method named getProfit that accepts profit from the user.
### Create a class Loss that has a method named getLoss that accepts loss from the user.
### Create a class BalanceSheet that inherits from both classes Profit and Loss and calculates the balanace. It has two methods getBalance() and printBalance().

```python
class Profit:
    def setProfit(self, profit):
        self.profit = profit

    def getProfit(self):
        return self.profit

class Loss:
    def setLoss(self, loss):
        self.loss = loss

    def getLoss(self):
        return self.loss

class BalanceSheet(Profit, Loss):
    def getBalance(self):
        return self.profit - self.loss

    def printBalance(self):
        print(f"Final Balance: {self.getBalance()}")

bs = BalanceSheet()

profit_amount = float(input("Enter profit amount: "))
loss_amount = float(input("Enter loss amount: "))

bs.setProfit(profit_amount)
bs.setLoss(loss_amount)

print(f"Profit: {bs.getProfit()}")
print(f"Loss: {bs.getLoss()}")
bs.printBalance()
```

### 15) WAP to demonstrate all types of inheritance.

```python
class Parent:
    def parent_method(self):
        print("This is Parent class")

class Child(Parent): 
    def child_method(self):
        print("This is Child class")

class Grandparent:
    def grandparent_method(self):
        print("This is Grandparent class")

class Parent2(Grandparent):
    def parent2_method(self):
        print("This is Parent2 class")

class Child2(Parent2): 
    def child2_method(self):
        print("This is Child2 class")

class Father:
    def father_method(self):
        print("This is Father class")

class Mother:
    def mother_method(self):
        print("This is Mother class")

class MultipleChild(Father, Mother):  
    def multiple_child_method(self):
        print("This is MultipleChild class")

class Base:
    def base_method(self):
        print("This is Base class")

class Derived1(Base):
    def derived1_method(self):
        print("This is Derived1 class")

class Derived2(Base):  
    def derived2_method(self):
        print("This is Derived2 class")

class A:
    def method_A(self):
        print("This is class A")

class B(A):
    def method_B(self):
        print("This is class B")

class C(A):
    def method_C(self):
        print("This is class C")

class D(B, C):  
    def method_D(self):
        print("This is class D")

print("\n Single Inheritance:")
obj1 = Child()
obj1.parent_method()
obj1.child_method()

print("\n Multilevel Inheritance:")
obj2 = Child2()
obj2.grandparent_method()
obj2.parent2_method()
obj2.child2_method()

print("\n Multiple Inheritance:")
obj3 = MultipleChild()
obj3.father_method()
obj3.mother_method()
obj3.multiple_child_method()

print("\n Hierarchical Inheritance:")
obj4 = Derived1()
obj5 = Derived2()
obj4.base_method()
obj4.derived1_method()
obj5.base_method()
obj5.derived2_method()

print("\n Hybrid Inheritance:")
obj6 = D()
obj6.method_A()
obj6.method_B()
obj6.method_C()
obj6.method_D()
     
```

### 16) Create a Person class with a constructor that takes two arguments name and age. 
### Create a child class Employee that inherits from Person and adds a new attribute salary. 
### Override the __init__ method in Employee to call the parent class's __init__ method using the super() and then initialize the salary attribute.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Employee(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary 

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Salary: {self.salary}")

name=input('Enter name of person:')
age=int(input('Enter age of person'))
salary=int(input('Enter salary of person'))
emp = Employee(name,age,salary)
emp.display()
```

### 17) Create a Shape class with a draw method that is not implemented. 
### Create three child classes Rectangle, Circle, and Triangle that implement the draw method with their respective drawing behaviors. 
### Create a list of Shape objects that includes one instance of each child class, and then iterate through the list and call the draw method on each object.

```python
from abc import ABC,abstractmethod
class Shape:
    @abstractmethod
    def draw(self):
        pass
class Rectangle(Shape):
    def draw(self):
        print("Drawing a Rectangle")

class Circle(Shape):
    def draw(self):
        print("Drawing a Circle")

class Triangle(Shape):
    def draw(self):
        print("Drawing a Triangle")

shapes = [Rectangle(), Circle(), Triangle()]
for shape in shapes:
    shape.draw()
```
