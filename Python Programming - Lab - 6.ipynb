# Tuple

### 01) WAP to find sum of tuple elements.

```python
a = int(input('Enter number of terms in tuple: '))
t1 = () 
for i in range(a):
    temp = int(input('Enter a number: '))
    t1 = t1 + (temp,)

sum_of_elements = 0
for i in t1:
    sum_of_elements += i

print(f'Sum of all elements of the tuple is: {sum_of_elements}')
```

### 03) WAP to find tuples which have all elements divisible by K from a list of tuples.

```python
l1 = [(1, 2, 3, -2, 3), (2, 4, 6), (4, 2, 21, 3234, 32), (-1, 23, 4), (3,)] 
k=int(input('Enter k: '))
ans=[]
for i in l1:
    positive=True
    for j in i:
        if j%k!=0:
            positive=False
            break
    if positive==True:
        ans.append(i)
print(ans)
```

### 09) WAP to multiply adjacent elements of a tuple and print that resultant tuple.

```python
a = int(input('Enter number of terms in tuple: '))
t1 = () 
for i in range(a):
    temp = int(input('Enter a number: '))
    t1 = t1 + (temp,)
ans=()
for i in range(0,len(t1)-1):
    ans=ans+(t1[i]*t1[i+1],)
print(ans)
```
